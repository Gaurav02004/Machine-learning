```python
from sklearn.datasets import fetch_california_housing

boston_df = fetch_california_housing(as_frame=True)

x = pd.DataFrame(boston_df.data, columns=boston_df.feature_names)
y = boston_df.target

x.head()
```

```python
##Independent features
x = pd.DataFrame(boston_df.data,columns = boston_df.feature_names)
y = boston_df.target
x.head()
```

```python
# Train Test Split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.33,
    random_state=42
)

print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)
print("Y_train shape:", y_train.shape)
print("Y_test shape:", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
regressor = DecisionTreeRegressor()
```

```python
regressor.fit(x_train,y_train)
```

```python
y_pred = regressor.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
```

```python
score = r2_score(y_pred,y_test)
```

```python
##Hyperparameter tunning
parameter = {
    'criterion':['squared_error','friedman_mse','absolute_error','poission'],
    'splitter': ['best','random'],
    'max_depth':[1,2,3,4,5,6,7,8,10,11,12],
    'max_features':['auto','sqrt','log2'],
}
```

```python
regressor = DecisionTreeRegressor()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
regressor = GridSearchCV(regressor,param_grid = parameter,cv=5,scoring = 'neg_mean_squared')
```

```python
from sklearn.model_selection import GridSearchCV

regressorcv = GridSearchCV(
    estimator=regressor,
    param_grid=parameter,
    cv=5,
    scoring='neg_mean_squared_error'
)
```

```python
# 1
from sklearn.model_selection import GridSearchCV

# 2
regressorcv = GridSearchCV(
    estimator=regressor,
    param_grid=parameter,
    cv=5,
    scoring='neg_mean_squared_error'
)

# 3
regressorcv.fit(x_train, y_train)

# 4
print("Best Parameters:", regressorcv.best_params_)

# 5
y_pred = regressorcv.predict(x_test)

# 6
from sklearn.metrics import r2_score
print("R2 Score:", r2_score(y_test, y_pred))
```
